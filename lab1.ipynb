```python
def task_8():
    raw_input_data = input().split()

    if not raw_input_data:
        return "Error. An empty string was entered."

    arr = []
    for item in raw_input_data:
        try:
            num = float(item) if '.' in item else int(item)
            arr.append(num)
        except ValueError:
            return 'Error. The list contains a non-numeric value.'

    idx = 0
    n = len(arr)
    while idx < n:
        if idx == 0 or arr[idx] <= arr[idx - 1]:
            idx += 1
        else:
            arr[idx], arr[idx - 1] = arr[idx - 1], arr[idx]
            idx -= 1

    return arr
```
